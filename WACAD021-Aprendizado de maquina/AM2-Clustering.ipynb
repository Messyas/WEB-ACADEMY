# Aprendizado Não Supervisionado - Agrupamento/Clustering (WebAcademy)

## Introdução: A Busca por Padrões Ocultos em Dados da Web

No desenvolvimento de aplicações web, a interação com dados estruturados e bem definidos é uma rotina. Tabelas em bancos de dados relacionais, como uma tabela `users` com uma coluna booleana `is_subscribed` ou uma tabela `products` com uma `category_id`, são a base de muitas funcionalidades. Este é o universo do Aprendizado Supervisionado, uma abordagem de machine learning onde um algoritmo aprende a partir de dados que já contêm as respostas corretas, ou "rótulos". A analogia mais próxima é a de aprender com cartões de memorização (*flashcards*): cada cartão possui uma pergunta (os dados de entrada) e uma resposta correta no verso (o rótulo). O objetivo é treinar um modelo que possa prever a resposta correta para novos cartões que ele nunca viu.

Contudo, a vasta maioria dos dados gerados na web não se encaixa nesse paradigma. Pense nos logs brutos de servidores, nos fluxos de cliques de usuários (clickstreams), nas avaliações de produtos, nos posts em redes sociais e nas imagens enviadas pelos usuários. Esses dados são massivos, desestruturados e, crucialmente, não vêm com uma "resposta correta" pré-definida. É aqui que entra o Aprendizado Não Supervisionado. Esta é a vertente do machine learning projetada para trabalhar com dados não rotulados, com o objetivo de descobrir padrões, estruturas e relações ocultas de forma autônoma, sem supervisão humana. A analogia aqui seria receber uma caixa cheia de peças de Lego de diversos tipos, sem um manual de instruções. A tarefa é organizar essas peças em grupos lógicos com base em suas características intrínsecas — cor, formato, tamanho. Não se sabe de antemão como esses grupos serão nomeados, mas é possível descobrir sua existência organicamente.

Essa transição do supervisionado para o não supervisionado representa uma mudança fundamental na mentalidade de resolução de problemas para um desenvolvedor. O foco deixa de ser a *previsão* ("qual é o rótulo correto para este dado?") e passa a ser a *descoberta* ("qual estrutura inerente existe nestes dados?"). A avaliação de sucesso de um modelo também muda. Em vez de medir a acurácia contra um gabarito conhecido, o sucesso de um modelo não supervisionado é frequentemente medido pela utilidade e pela capacidade de ação dos insights que ele gera para o negócio. A pergunta-chave não é "O modelo está certo?", mas sim "Os padrões que o modelo encontrou são úteis e significativos?".

### A Tarefa Central do Clustering: Encontrando Grupos Naturais

Uma das tarefas mais fundamentais e poderosas no aprendizado não supervisionado é o *clustering*, ou agrupamento. O objetivo do clustering é particionar um conjunto de dados em subgrupos, conhecidos como *clusters*, de tal forma que os pontos de dados dentro de um mesmo cluster sejam muito semelhantes entre si, enquanto os pontos de dados em clusters diferentes sejam muito distintos. Imagine um gráfico de dispersão com pontos de dados não rotulados que, visualmente, formam "nuvens" ou aglomerações distintas. A tarefa do algoritmo de clustering é, essencialmente, desenhar fronteiras ao redor dessas nuvens de forma programática, revelando a estrutura de grupos que existe nos dados.

## O Poder do Agrupamento: Segmentação de Clientes em Aplicações Web

Para um desenvolvedor web, a capacidade de encontrar grupos naturais nos dados abre um leque de possibilidades para criar aplicações mais inteligentes, personalizadas e eficientes. A segmentação de clientes é, talvez, a aplicação mais direta e impactante do clustering no ecossistema web.

### Segmentando Usuários por Comportamento: De Cliques a Personas

Tradicionalmente, a segmentação de usuários é feita com base em regras de negócio predefinidas, como `IF user_purchases > 5 THEN segment = 'loyal'`. Embora útil, essa abordagem é limitada pelas suposições humanas. O clustering, por outro lado, permite que os próprios dados revelem as divisões mais significativas e naturais entre os usuários.

**Exemplo Prático: E-commerce**

1.  **Coleta de Dados Não Rotulados:** Uma aplicação de e-commerce coleta uma vasta quantidade de dados comportamentais brutos para cada usuário, como: duração da sessão, páginas visitadas por sessão, número de itens adicionados ao carrinho, taxa de conversão de compras, valor monetário total gasto, dias desde a última visita (métricas de Recência, Frequência e Valor Monetário, ou RFM), e tipo de dispositivo usado.
2.  **Descoberta de Clusters:** Um algoritmo de clustering pode processar esses dados multidimensionais e agrupar os usuários em clusters distintos.
3.  **Criação de Personas:** A etapa final é a interpretação desses clusters para criar "personas" acionáveis, que são arquétipos de clientes baseados em comportamento real. Por exemplo, o algoritmo pode identificar os seguintes grupos:
    * **Cluster 1: "Campeões de Alto Valor"**: Usuários com alta frequência de compra, alto valor monetário e compras recentes.
    * **Cluster 2: "Compradores Indecisos"**: Usuários que adicionam muitos itens ao carrinho, mas têm uma baixa taxa de conversão, passando muito tempo no site.
    * **Cluster 3: "Caçadores de Ofertas"**: Usuários que visitam o site com frequência durante promoções, possuem um valor médio de pedido baixo e respondem bem a cupons de desconto.
    * **Cluster 4: "Novatos Curiosos"**: Usuários com poucas visitas, uma primeira compra recente e que navegam por diversas categorias.

O impacto de negócio dessa segmentação é imenso. Em vez de enviar a mesma campanha de marketing para todos, é possível criar ações personalizadas. Os "Compradores Indecisos" podem receber um e-mail de abandono de carrinho com um incentivo, enquanto os "Campeões de Alto Valor" podem ser convidados para um programa de fidelidade exclusivo. Estudos de caso demonstram que essa personalização baseada em segmentos aumenta drasticamente métricas de engajamento, como a Taxa de Cliques (CTR) e as conversões.

### Agrupamento de Produtos Similares: Aprimorando Recomendações e Busca

O clustering não se limita a usuários; ele também pode organizar o catálogo de produtos de uma aplicação. Em vez de depender apenas de dados de vendas ou categorias manuais, o clustering pode agrupar produtos com base em seus atributos e metadados.

**Exemplo Prático: Serviço de Streaming ou E-commerce de Moda**

1.  **Coleta de Dados de Atributos:** Para um filme, as características podem incluir vetores de gênero, diretor, elenco principal e palavras-chave extraídas da sinopse. Para uma peça de roupa, as características podem ser cor, estilo, tipo de tecido e estação.
2.  **Descoberta de Clusters de Produtos:** O clustering dessas características agruparia "vestidos de verão florais e leves" em um cluster, bem separado de "casacos de lã pesados para o inverno" em outro.
3.  **Aplicações Práticas:**
    * **Recomendações Mais Inteligentes:** Quando um usuário visualiza um produto, o sistema pode recomendar outros itens do mesmo cluster. Isso constitui uma forma de filtragem baseada em conteúdo, que é extremamente eficaz.
    * **Solução para o "Problema do Início Frio" (*Cold Start*):** Um produto novo no catálogo não possui histórico de interações (visualizações, compras). Os sistemas de recomendação tradicionais (filtragem colaborativa) falham nesse cenário. No entanto, ao analisar os atributos do novo produto, podemos inseri-lo em um cluster existente. Imediatamente, ele pode ser recomendado a usuários que já demonstraram interesse em outros produtos daquele cluster, resolvendo um problema crítico de negócio.

## O Algoritmo K-Means: Um Mergulho Profundo para Desenvolvedores

Um dos algoritmos de clustering mais populares e fundamentais é o K-Means. Sua popularidade deriva de sua simplicidade, eficiência e interpretabilidade.

### A Intuição: Uma Dança de Centróides

O K-Means é um algoritmo iterativo baseado em centróides. Seu objetivo é minimizar a *inércia*, também conhecida como Soma dos Quadrados Intra-Cluster (WCSS - *Within-Cluster Sum of Squares*). Essa métrica representa a soma das distâncias ao quadrado entre cada ponto de dados e o centro (centróide) do seu cluster atribuído. Essencialmente, o algoritmo tenta criar os clusters mais compactos e densos possíveis.

O processo, conhecido como algoritmo de Lloyd, pode ser visualizado como uma "dança" dos centróides até que encontrem suas posições ideais. O processo ocorre nos seguintes passos:

1.  **Passo 1: Inicialização.** O primeiro passo é definir o número de clusters que se deseja encontrar, um hiperparâmetro denotado por $K$. Em seguida, o algoritmo posiciona $K$ "centróides" iniciais em locais aleatórios no espaço de dados.
2.  **Passo 2: Atribuição.** Cada ponto de dados no conjunto é atribuído ao centróide mais próximo. A "proximidade" é geralmente medida pela distância Euclidiana (a distância em linha reta entre dois pontos). Esta etapa cria as primeiras $K$ partições nos dados, que podem ser visualizadas como um diagrama de Voronoi, onde cada região contém os pontos mais próximos de um determinado centróide.
3.  **Passo 3: Atualização.** Após todos os pontos terem sido atribuídos a um cluster, a posição de cada um dos $K$ centróides é recalculada. A nova posição do centróide é a *média* de todos os pontos de dados que foram atribuídos a ele na etapa anterior. É daí que vem o termo "Means" (médias) no nome do algoritmo.
4.  **Passo 4: Repetição.** Os passos 2 e 3 são repetidos iterativamente. Na nova etapa de atribuição, alguns pontos de dados podem agora estar mais próximos de um centróide diferente (que se moveu na etapa de atualização) e, portanto, mudam de cluster. Isso é seguido por outra atualização dos centróides.
5.  **Convergência.** Este processo iterativo de atribuição e atualização continua até que os centróides parem de se mover significativamente ou até que as atribuições dos pontos aos clusters não mudem mais. Neste ponto, o algoritmo convergiu, e os clusters finais estão formados.

### Um Passo de Pré-processamento Inegociável: A Importância da Normalização

O K-Means é um algoritmo baseado em distância e, por isso, é extremamente sensível à escala das *features* (variáveis) nos dados. Se as features tiverem escalas ou variâncias muito diferentes, aquelas com magnitudes maiores irão dominar o cálculo da distância, efetivamente anulando a contribuição das outras.

**Exemplo Prático: Idade vs. Salário**

Imagine um conjunto de dados de clientes com duas features: `Idade` (variando de 18 a 70) e `Salario_Anual` (variando de R$ 30.000 a R$ 150.000). Ao calcular a distância Euclidiana ao quadrado:
* Uma diferença de 20 anos na idade (ex: 25 vs 45) contribui com $20^2 = 400$ para a soma.
* Uma diferença de R$ 20.000 no salário contribui com $20000^2 = 400.000.000$ para a soma.

Claramente, a feature `Salario_Anual` irá dominar completamente a formação dos clusters. O algoritmo agrupará os clientes quase que exclusivamente com base no salário, ignorando a idade.

A solução para este problema é a **normalização** ou **padronização** dos dados. Esta é uma técnica de pré-processamento que transforma todas as features para uma escala comum, garantindo que todas tenham o mesmo peso no cálculo da distância. O método mais comum é a Padronização (*Standardization*), que reescala os dados para que tenham uma média de 0 e um desvio padrão de 1. Na biblioteca Scikit-learn, isso é facilmente realizado com a classe `StandardScaler`.

A simplicidade do K-Means é tanto sua maior força quanto sua principal fraqueza. Suas suposições intrínsecas — de que os clusters são esféricos (isotrópicos) e de tamanho semelhante — o tornam extremamente rápido e escalável, o que é uma grande vantagem para aplicações web com milhões de usuários. No entanto, se os clusters reais nos dados tiverem formatos alongados, não convexos (como luas crescentes) ou densidades muito diferentes, o K-Means falhará em identificá-los corretamente. Isso significa que a escolha do K-Means implica uma suposição sobre a geometria dos padrões nos dados. Se essa suposição estiver errada, os resultados podem ser enganosos, tornando crucial a visualização dos dados e a consideração de algoritmos alternativos, como o DBSCAN, para estruturas de cluster mais complexas.

```python
# Célula de código para pré-processamento e normalização dos dados
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_features = scaler.fit_transform(df[['Idade', 'Salario_Anual']])
```

## De Clusters a Insights: Dando Sentido aos Grupos

Após a execução do algoritmo, o trabalho não está concluído. A etapa mais crítica é a interpretação dos clusters para extrair insights acionáveis.

### O Dilema do "Quantos Grupos?": Escolhendo o K Ótimo

O maior desafio prático do K-Means é a necessidade de especificar o número de clusters, $K$, antecipadamente. Uma escolha inadequada de $K$ pode levar a agrupamentos que não representam a estrutura real dos dados. Felizmente, existem métodos heurísticos para guiar essa escolha.

**Método 1: O Método do Cotovelo (*Elbow Method*)**

Este método consiste em executar o algoritmo K-Means para um intervalo de valores de $K$ (por exemplo, de 1 a 10) e, para cada valor, calcular a métrica WCSS (inércia).
* **Conceito:** A WCSS sempre diminuirá à medida que $K$ aumenta. Se $K$ for igual ao número de pontos de dados, a WCSS será zero. O objetivo é encontrar o ponto em que a diminuição da WCSS se torna marginal.
* **Visualização e Interpretação:** Ao plotar $K$ no eixo x e a WCSS no eixo y, o gráfico geralmente se assemelha a um braço. O "cotovelo" dessa curva — o ponto onde a taxa de queda diminui drasticamente — é considerado o valor ótimo para $K$. Este ponto representa o melhor equilíbrio entre a compactação dos clusters e a simplicidade do modelo (não ter clusters demais).

![image.png](https://medium.com/@zalarushirajsinh07/the-elbow-method-finding-the-optimal-number-of-clusters-d297f5aeb189)

**Método 2: Análise de Silhueta (*Silhouette Analysis*)**

Este é um método mais robusto que avalia a qualidade dos clusters medindo quão bem cada ponto se encaixa em seu próprio cluster em comparação com outros clusters. A análise considera dois fatores para cada ponto:
* **Coesão ($a$):** A distância média para todos os outros pontos no mesmo cluster.
* **Separação ($b$):** A distância média para todos os pontos no cluster mais próximo.

O Coeficiente de Silhueta para um único ponto é calculado como $S = (b - a) / max(a, b)$ e varia de -1 a +1.
* **+1:** Indica que o ponto está bem ajustado ao seu cluster e longe dos clusters vizinhos.
* **0:** Indica que o ponto está próximo da fronteira entre dois clusters.
* **-1:** Indica que o ponto pode ter sido atribuído ao cluster errado.

Para encontrar o $K$ ótimo, calcula-se a pontuação média da silhueta para diferentes valores de $K$. O valor de $K$ que maximiza essa pontuação média é geralmente a melhor escolha.

```python
# Célula de código para implementar o Método do Cotovelo e a Análise de Silhueta
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

wcss = []
for i in range(1, 11):
     kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
     kmeans.fit(scaled_features)
     wcss.append(kmeans.inertia_)
 plt.plot(range(1, 11), wcss)
 plt.title('Método do Cotovelo')
 plt.xlabel('Número de Clusters (K)')
 plt.ylabel('WCSS')
 plt.show()
```

### Visualizando os Clusters para um Entendimento Mais Profundo

A visualização é uma ferramenta poderosa para validar e entender os clusters. Para dados com apenas duas features, um simples gráfico de dispersão, onde cada ponto é colorido de acordo com seu rótulo de cluster, é suficiente.

No entanto, a maioria dos problemas do mundo real envolve dados de alta dimensionalidade (muitas features). Nesses casos, a visualização direta não é possível. A solução é usar uma técnica de redução de dimensionalidade, como a **Análise de Componentes Principais (PCA)**.
* **Conceito:** A PCA transforma as features originais de alta dimensão em um novo conjunto de features de baixa dimensão, chamadas "componentes principais". Essas novas features são combinações das originais e são ordenadas pela quantidade de variância dos dados que elas capturam.
* **Aplicação:** Podemos reduzir as múltiplas features do nosso conjunto de dados para apenas os dois primeiros componentes principais, que capturam a maior parte da variância. Em seguida, podemos criar um gráfico de dispersão 2D usando esses dois componentes como eixos x e y, e colorir os pontos com base em seus rótulos de cluster do K-Means. Isso nos permite "ver" a separação dos clusters que foi encontrada no espaço de alta dimensão, fornecendo uma validação visual da qualidade do agrupamento.
