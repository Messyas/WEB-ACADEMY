### 1. Data Storytelling: Transformando Dados em Narrativas Impactantes

#### Análise Exploratória vs. Análise Explicatória: O Propósito de Cada Uma

No campo da análise de dados, é crucial distinguir entre duas abordagens fundamentais: a análise exploratória e a análise explanatória. Cada uma possui um propósito distinto e é aplicada em diferentes estágios do processo analítico.

* **Análise Exploratória**:
    * É o processo de familiarização com os dados.
    * Pode-se iniciar com uma hipótese ou pergunta, ou simplesmente mergulhar nos dados para descobrir o que pode ser interessante.
    * Frequentemente comparada a "virar 100 pedras para encontrar talvez 1 ou 2 joias preciosas".
    * Envolve alta interatividade, permitindo ao usuário manipular visualizações e dados para desvendar suas próprias descobertas.
    * Pode conter uma grande quantidade de informações.

* **Análise Explicatória**:
    * Ocorre quando já se tem algo específico para apresentar a uma audiência – geralmente as "joias preciosas" identificadas na fase exploratória.
    * O objetivo é contar uma história clara, guiando a audiência para descobertas específicas e respondendo a perguntas predefinidas.
    * A ordem das visualizações é pensada para conduzir o usuário através de uma narrativa que culmina na resposta a essas perguntas, pintando um quadro claro para a audiência.
    
---

* A análise de dados segue uma sequência lógica: primeiro a **exploratória** (a fase de descoberta), depois a **explanatória** (a fase de comunicação).

* Realizar a análise explanatória *após* a exploratória é fundamental para focar a atenção da audiência nos insights mais relevantes.

* Essa prática evita a apresentação de um volume desorganizado de dados, favorecendo uma narrativa coesa e clara.

* Para desenvolvedores, isso significa que dashboards, embora possam ter elementos interativos (filtros), devem ser projetados com uma **intenção explanatória** principal.

* O desafio central é equilibrar a capacidade de exploração do usuário com a necessidade de transmitir uma mensagem clara e inequívoca.

A tabela a seguir resume as principais diferenças entre as duas abordagens:

| Característica | Análise Exploratória | Análise Explicatória |
| :-------------------- | :--------------------------------------------------------- | :----------------------------------------------------------- |
| **Propósito** | Familiarizar-se com os dados, descobrir padrões e hipóteses. | Comunicar descobertas específicas e conclusões para uma audiência. |
| **Audiência** | O próprio analista, equipe de dados. | Stakeholders, gerentes, clientes, público em geral. |
| **Tipo de Pergunta** | Aberta, "O que os dados me dizem?" | Específica, "Qual é a resposta para esta pergunta de negócio?" |
| **Interatividade** | Alta (filtros, drill-downs, visualizações dinâmicas). | Controlada (guiada por uma narrativa, com interatividade para aprofundamento). |
| **Volume de Dados** | Geralmente alto, muitos gráficos e visualizações. | Seletivo, focado nas "joias preciosas" descobertas. |
| **Foco** | Descoberta e entendimento. | Clareza, persuasão e chamada para ação. |

#### A Estrutura de uma Narrativa: Contexto, Trama (Insights), Conclusão/Chamada para Ação

Uma história de dados eficaz adota uma estrutura narrativa familiar, análoga à de qualquer boa história, para maximizar seu impacto e memorabilidade. Essa estrutura é composta por três elementos chave:

1.  **Contexto (Início)**:
    * Esta é a introdução, onde se estabelece o cenário e se define o problema de negócio ou a oportunidade que está sendo abordada.
    * **Exemplo**: "A diretoria deseja compreender o perfil dos nossos usuários para otimizar as estratégias de marketing."

2.  **A "Trama" / Conflito (Meio)**:
    * Nesta parte, são reveladas as descobertas obtidas a partir dos dados, que explicam o que está acontecendo.
    * **Exemplo**: "Constatamos que a maioria dos nossos usuários de alto valor reside na região Sudeste e se cadastrou nos últimos seis meses."

3.  **Conclusão / Chamada para Ação (Fim)**:
    * O desfecho da história apresenta a recomendação baseada nas descobertas.
    * Indica o que precisa ser feito a seguir.
    * **Exemplo**: "Sugerimos direcionar campanhas de marketing digital específicas para essa região."

### O Poder da Narrativa na Análise de Dados

* **Impacto Neurológico:** A narrativa engaja múltiplas áreas do cérebro, ativando a compreensão (área de Wernicke), a emoção (amígdala) e a empatia (neurônios-espelho).
* **Além dos Fatos:** Uma história de dados não apenas apresenta informações, mas busca **persuadir e motivar**, tornando os insights mais memoráveis e acionáveis.
* **Estrutura Comprovada:** O modelo de **"contexto, conflito e resolução"** é uma forma eficaz de organizar a narrativa e guiar a audiência.
* **Implicação para Desenvolvedores:** Dashboards não devem ser repositórios de dados, mas sim ferramentas que contam uma história, com um fluxo lógico que leva o usuário do problema à solução. A interatividade pode ser usada para aprofundar a "trama".

#### Conhecendo sua Audiência: Adaptando a Mensagem para Diferentes Stakeholders

A eficácia de uma história de dados depende diretamente de sua adaptação à audiência. É fundamental ajustar a linguagem e a complexidade da narrativa para quem irá consumir a informação, seja um CEO, um gerente de marketing, ou outros desenvolvedores.

Ao preparar uma história de dados, é importante considerar os seguintes aspectos da audiência:

* **Necessidades e Interesses**: Quais informações são mais relevantes para o papel e os objetivos específicos de cada stakeholder?
* **"Dor" (o que está em jogo)**: Quais desafios ou oportunidades impactam diretamente a audiência e quais são as implicações para eles?
* **Atitudes**: A audiência já confia no apresentador, ou a confiança precisa ser conquistada? Como eles podem reagir ou resistir à ação proposta?
* **Papéis Organizacionais e Níveis de Autoridade**: Esses fatores influenciam o nível de detalhe e o tipo de informação que será mais apreciado.
* **Conhecimento de Domínio e Experiência**: A complexidade dos dados e das explicações deve ser adaptada ao nível de compreensão técnica da audiência.
* **Crenças e Opiniões Enraizadas, Demografia**: Essas nuances podem influenciar a forma como a mensagem é recebida.

### A Adaptação da Narrativa como Chave para o Impacto

* **O Princípio Fundamental:** A adaptação da história para a audiência é um imperativo estratégico para gerar impacto, confiança e levar à ação.

* **Estudo de Caso: Beth na "Play On"**
    * A analista Beth utilizou os **mesmos dados** para criar **duas narrativas distintas** e persuadir públicos diferentes.

    * **Público 1: Chief Marketing Officer (CMO)**
        * **Foco da Narrativa:** Apelo do produto e encantamento infantil.
        * **Abordagem:** Usou gráficos combinados com fotos de crianças felizes com o brinquedo para ressoar com as metas de marketing e engajamento.

    * **Público 2: CEO**
        * **Foco da Narrativa:** Ameaça de mercado e deterioração financeira.
        * **Abordagem:** Posicionou uma startup concorrente como antagonista, destacando a rápida erosão da participação de mercado para apelar às preocupações estratégicas do CEO.

* **A Lição Principal:** A falha em adaptar a mensagem pode resultar em desconfiança ou fazer com que a análise pareça irrelevante para a audiência.

* **Implicação para Desenvolvedores:**
    * Dashboards podem precisar de **camadas de informação ou visualizações diferentes** para atender a diversas audiências.
    * A apresentação verbal de um dashboard deve ser flexível, destacando os pontos mais relevantes para cada grupo específico.

**Exemplos de Adaptação para Diferentes Audiências**:

* **Para o CEO**: O foco deve ser em Indicadores Chave de Performance (KPIs) de alto nível, implicações estratégicas, posição de mercado, riscos e Retorno sobre o Investimento (ROI). A história deve ser concisa e orientada para a decisão.
* **Para o Gerente de Marketing**: A narrativa deve se concentrar em segmentos de clientes, desempenho de campanhas, taxas de conversão e engajamento do usuário. A história precisa ser acionável para a otimização de campanhas.
* **Para Outros Desenvolvedores**: A atenção deve ser dada à viabilidade técnica, qualidade dos dados, desempenho do sistema, desafios de implementação e oportunidades de inovação. Essa audiência apreciará detalhes técnicos e a lógica por trás das escolhas de dados e ferramentas.

## 2. Dashboards Web: Visão Geral e Boas Práticas de Design

### O que é um Dashboard? Definição e Propósito

* **Definição Principal:** Uma interface visual projetada para fornecer, de relance, as informações mais importantes necessárias para atingir um ou mais objetivos.

* **Função Primordial:** Apresentar dados complexos de maneira clara e concisa, permitindo que os usuários tomem decisões informadas rapidamente.

* **Além da Exibição:** Dashboards eficazes não apenas apresentam dados; eles devem **transmitir a história** por trás deles, orientando o usuário sem sobrecarregá-lo com detalhes excessivos.

---

### O Propósito Estratégico de um Dashboard

* **Ferramenta de Decisão:** Um dashboard é uma ferramenta para tomar decisões, não apenas um exibidor de dados. Seu propósito é transformar dados em **valor de negócio**.

* **Design Orientado ao Usuário:** O design deve ser impulsionado pelo propósito e pelas necessidades específicas do usuário. Um dashboard que apenas "despeja dados" sem estrutura ou contexto falha em seu objetivo.

* **Desafio de UX:** A construção de um dashboard é um exercício de **Design de Experiência do Usuário (UX)** focado em facilitar a tomada de decisão.

* **Clareza de Propósito:** Cada elemento visual no dashboard deve ter um propósito claro e contribuir para o objetivo final do usuário.

#### Princípios Essenciais de Design de Dashboards: Clareza, Hierarquia, Contexto, Escolha de Gráficos

Para que um dashboard seja verdadeiramente eficaz, ele deve aderir a um conjunto de princípios de design que garantam sua usabilidade e impacto.

* **Clareza e Simplicidade**:
    * O princípio "menos é mais" é fundamental. É preciso evitar a poluição visual e o excesso de informações, garantindo que o dashboard seja compreensível à primeira vista.
    * Dashboards devem ser projetados para transmitir dados rapidamente, evitando aglomeração e texto excessivo.
    * O uso estratégico de espaço em branco (whitespace) é crucial para permitir que o conteúdo "respire", tornando-o mais fácil de ler e compreender.

* **Hierarquia da Informação**:
    * Os Indicadores Chave de Performance (KPIs) mais importantes devem ser colocados no topo ou em uma posição de destaque, utilizando recursos visuais como tamanho, cor e posição para atrair a atenção.
    * A hierarquia visual é essencial para guiar a atenção do usuário para as informações mais críticas primeiramente.
    * Dada a tendência de leitura da esquerda para a direita e de cima para baixo, as informações cruciais devem ser posicionadas no canto superior esquerdo do dashboard.

* **Contexto é Rei**:
    * Um número isolado carece de significado. É imperativo compará-lo com uma meta, um período anterior, uma média ou um benchmark para conferir-lhe relevância.
    * A ausência de contexto impede a interpretação da significância de um dado.

* **Escolha Certa dos Gráficos**:
    * A seleção do tipo de gráfico apropriado para a pergunta e o tipo de dado é vital (revisão da Aula 2).
    * **Gráficos de barras** são excelentes para comparações entre categorias.
    * **Gráficos de linha** são ideais para exibir tendências ao longo do tempo.
    * **Gráficos de pizza (pie charts)** são mais adequados para representar 2 a 3 categorias que compõem um todo; deve-se evitar o uso com muitos segmentos.
    * **Cuidado**: Gráficos de barras devem sempre ter seu eixo Y iniciando em zero para evitar distorções na percepção dos dados.

* **Interatividade**:
    * Um dashboard eficaz deve ser interativo, permitindo que os usuários manipulem os dados (por exemplo, filtrando e classificando) para atender às suas necessidades específicas.
    * Recursos como filtros, capacidades de drill-down e visualizações personalizadas permitem que os usuários explorem os dados em seu próprio ritmo.

* **Consistência Visual**:
    * Manter a uniformidade no uso de cores, fontes, rótulos e layout é crucial. A consistência no design facilita a compreensão e a navegação do usuário.
    * Um design claro é assegurado pela adesão consistente à tipografia, cores e elementos gráficos.

### Minimizando a Carga Cognitiva: O Design Inteligente

* **O Princípio Fundamental:** Um dos objetivos mais importantes no design de dashboards é a **minimização da carga cognitiva** para o usuário.

* **O Efeito de um Design Ruim:** Um layout complexo, confuso ou excessivamente carregado leva à sobrecarga, frustração e, por fim, à incapacidade de extrair conclusões significativas.

* **O Objetivo Final:** O usuário deve ser capaz de interpretar os dados "de relance" e "rapidamente", sem esforço desnecessário.

* **Conselho para Desenvolvedores:** A tentação de adicionar "tudo" ou de usar visuais apenas por serem chamativos deve ser resistida. A prioridade é sempre a **clareza e o propósito**.

* **A Regra de Ouro:** Simplicidade e intencionalidade no design são muito mais valiosas do que a complexidade visual.

#### Erros Comuns a Evitar no Design de Dashboards

Apesar dos princípios de design, é comum observar certos erros que comprometem a eficácia de um dashboard. Evitá-los é tão importante quanto aplicar as boas práticas.

* **Sobrecarga de Informação**: Tentar incluir demasiada informação em uma única visualização ou no dashboard como um todo. Isso resulta em "poluição visual" e dificulta a extração de conclusões.
* **Falta de Contexto**: Apresentar números sem qualquer comparação ou ponto de referência (como uma meta, dados de um período anterior, ou um benchmark) torna os dados sem significado.
* **Uso Excessivo de Cores**: Embora as cores possam realçar um dashboard, o uso indiscriminado e excessivo pode ser distrativo e confuso. As cores devem ser utilizadas intencionalmente para destacar ou categorizar informações.
* **Design Muito Ocupado/Distrativo**: Designs "chamativos", elementos 3D ou gráficos excessivamente complexos que não adicionam valor podem distorcer a percepção dos dados e dificultar a compreensão.
* **Escolha Inadequada de Gráficos**: Utilizar um tipo de gráfico incorreto para o tipo de dado ou para a pergunta que se deseja responder.
* **Dados Mal Rotulados**: Rótulos insuficientes, imprecisos ou confusos levam a interpretações errôneas e confusão.
* **Não Otimizado para Mobile**: Ignorar a responsividade do dashboard para dispositivos móveis é um erro significativo no cenário atual.

### Erros Comuns: Priorizando a Estética sobre a Clareza

* **Exemplos de Erros:** "Design Muito Ocupado", "Muitas Cores" e "Gráficos de Pizza 3D" são falhas comuns que nascem do mesmo problema.

* **A Causa Principal:** Esses erros ocorrem quando a **estética é priorizada em detrimento da funcionalidade** e da clareza da informação.

* **A Relação Causal:** Designs visualmente atraentes, mas disfuncionais, levam diretamente à confusão, comprometendo o propósito da visualização e impedindo conclusões eficazes.

* **Lembrete para Desenvolvedores:** A interface deve servir ao propósito de comunicar dados, não ser um fim em si mesma. **Simplicidade e clareza** são as verdadeiras medidas de um bom design.



**Exemplos de Dashboards: Boas Práticas vs. Erros Comuns (Conceitual)**

Para ilustrar os princípios e erros discutidos, considere os seguintes exemplos conceituais de design de dashboards:

* **Exemplo de Dashboard Bem Projetado**:

    * Um dashboard limpo e organizado, com KPIs grandes e claramente visíveis.
    * Gráficos de linha são utilizados para mostrar tendências ao longo do tempo, enquanto gráficos de barras são empregados para comparações entre categorias.
    * A paleta de cores é limitada e consistente, com cores usadas intencionalmente para destacar informações importantes.
    * Todos os elementos são claramente rotulados, e os números são apresentados com contexto (por exemplo, setas de delta indicando mudança, ou comparação com uma meta).
    * Há um uso adequado de espaço em branco para evitar a sobrecarga visual.

* **Exemplos de Dashboard Mal Projetado**:



    * Um dashboard visualmente lotado, com muitos gráficos de tipos diferentes espremidos na mesma tela.
    * Uso excessivo de cores aleatórias, sem um propósito claro, tornando o dashboard confuso.
    * Inclusão de gráficos 3D que distorcem a percepção dos dados e dificultam a comparação.
    * Falta de rótulos claros ou rótulos confusos, levando a interpretações errôneas.
    * Números apresentados sem contexto, tornando impossível para o usuário determinar se um valor é bom ou ruim.

* **Ciclo de Desenvolvimento Rápido**:
    * Permite criar e iterar rapidamente em aplicações web interativas utilizando Python puro.
    * Elimina a necessidade de código boilerplate, permitindo que os usuários escrevam um script, o executem e vejam os resultados instantaneamente em um formato web.
    * A reflexão em tempo real das alterações no código reduz drasticamente o tempo do conceito à produção, o que é particularmente benéfico para projetos orientados a dados que exigem atualizações e visualizações frequentes.

    * Uma das grandes vantagens do Streamlit é a configuração mínima necessária para começar a criar aplicações.
    * Ao contrário de frameworks de desenvolvimento web mais complexos que exigem extensa configuração de ambiente e uma curva de aprendizado acentuada, o Streamlit simplifica o processo.
    * Desenvolvedores podem construir aplicações web funcionais com apenas algumas linhas de código Python, contornando os obstáculos usuais do desenvolvimento web, como lidar com servidores web, frameworks front-end ou as complexidades do ciclo de requisição HTTP.

* **Interfaces Altamente Interativas**:
    * Streamlit se destaca na criação de interfaces altamente interativas por meio de widgets integrados, como sliders, botões e caixas de seleção.
    * Esses elementos podem ser incorporados de forma contínua nas aplicações com código mínimo.
    * Essa interatividade permite que os usuários manipulem variáveis e vejam imediatamente o impacto nas exibições de dados ou saídas de modelos, proporcionando uma experiência de usuário intuitiva e envolvente.

* **Integração Perfeita com o Ecossistema Python**:
    * A integração do Streamlit com o ecossistema Python é uma vantagem significativa para desenvolvedores e cientistas de dados.

    * Sendo nativo de Python, o Streamlit se encaixa perfeitamente nos fluxos de trabalho de análise de dados existentes, eliminando a necessidade de converter scripts ou se adaptar a novos conjuntos de ferramentas.

* **Open Source com Desenvolvimento Ativo e Suporte da Comunidade**:

* **Deployment Simples**:
    * O Streamlit oferece um processo de deployment direto, simplificando a transição do desenvolvimento para a produção.
    * As aplicações são fáceis de configurar e implantar com configuração mínima, um benefício significativo para compartilhar aplicações com um público mais amplo sem as complexidades típicas de deployment de aplicações web.

### Streamlit: Democratizando a Criação de Aplicações de Dados

**Código para Geração de `usuarios_limpo.csv` (execute esta célula para criar o arquivo)**:

````python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random
import os

def generate_synthetic_user_data(num_users=5000):
    np.random.seed(42)
    random.seed(42)

    user_ids = [f'user_{i:05d}' for i in range(num_users)]
    states = ['SP', 'MG', 'RJ', 'BA', 'PR', 'RS', 'SC', 'GO', 'DF', 'PE', 'CE', 'AM']
    # Definir probabilidades para ter uma distribuição mais realista
    probabilities = [0.25, 0.15, 0.15, 0.05, 0.1, 0.08, 0.07, 0.05, 0.05, 0.02, 0.02, 0.01]
    # Garantir que as probabilidades somem 1
    probabilities = [p / sum(probabilities) for p in probabilities]
    user_states = np.random.choice(states, num_users, p=probabilities)

    end_date = datetime.now()
    start_date = end_date - timedelta(days=2*365) # Últimos 2 anos
    registration_dates = [start_date + timedelta(days=random.randint(0, (end_date - start_date).days)) for _ in range(num_users)]
    registration_dates.sort() # Para simular uma ordem cronológica

    last_purchase_values = np.random.lognormal(mean=5.5, sigma=0.8, size=num_users)
    last_purchase_values = np.round(last_purchase_values, 2)
    # Ajustar valores extremos para serem mais realistas
    last_purchase_values[last_purchase_values > 1000] = 1000 + np.random.rand(np.sum(last_purchase_values > 1000)) * 500
    last_purchase_values[last_purchase_values < 10] = 10 + np.random.rand(np.sum(last_purchase_values < 10)) * 20

    df = pd.DataFrame({
        'user_id': user_ids,
        'estado': user_states,
        'data_cadastro': registration_dates,
        'valor_ultima_compra': last_purchase_values
    })
    df['data_cadastro'] = pd.to_datetime(df['data_cadastro']) # Garantir tipo datetime
    return df

# Gerar e salvar o arquivo CSV
usuarios_df = generate_synthetic_user_data(num_users=5000)
file_path = 'usuarios_limpo.csv'
usuarios_df.to_csv(file_path, index=False)
print(f"Arquivo '{file_path}' gerado com sucesso no diretório atual!")
print("Primeiras 5 linhas do DataFrame:")
print(usuarios_df.head())
print("\nInformações do DataFrame:")
print(usuarios_df.info())
````

#### Adicionando um Filtro de Seleção Múltipla para os Estados com `st.sidebar.multiselect()`

A função `st.multiselect()` permite que o usuário selecione uma ou várias opções de uma lista, proporcionando flexibilidade na filtragem de dados.

#### Lógica de Filtragem Dinâmica do DataFrame

O DataFrame principal (`df`) será filtrado dinamicamente com base nas seleções do usuário na barra lateral.

A funcionalidade central do Streamlit reside em seu modelo de programação reativa: o aplicativo é "automaticamente executado novamente por completo quando os usuários interagem com os widgets". Isso simplifica drasticamente a construção de interfaces interativas, pois o desenvolvedor não precisa gerenciar explicitamente o estado ou callbacks complexos. Quando o filtro `st.multiselect` é alterado, o script inteiro é reexecutado, o DataFrame é filtrado novamente, e todos os KPIs e gráficos são atualizados automaticamente. Para desenvolvedores web acostumados com manipulação de DOM e gerenciamento de estado em JavaScript, essa abstração é um diferencial significativo. Permite focar na lógica de dados e visualização, em vez da complexidade da interface. O desafio pode ser a otimização de performance para grandes datasets, onde o uso de cache (`@st.cache_data`) se torna essencial para evitar reprocessamentos desnecessários.

**Exemplo de Código para Lógica de Filtragem (Parte do seu `app.py`)**:

````python

# Aplicar a lógica de filtragem ao DataFrame
if estados_selecionados: # Verificar se há estados selecionados
    df_filtrado = df[df['estado'].isin(estados_selecionados)]
else:
    df_filtrado = df # Se nenhum estado for selecionado, mostrar todos os dados
    st.sidebar.warning("Nenhum estado selecionado. Exibindo dados de todos os estados.")
````

#### Fazer os KPIs e Gráficos se Atualizarem Dinamicamente

Para que os KPIs e gráficos reflitam as seleções do filtro, basta utilizar o `df_filtrado` (o DataFrame resultante da filtragem) em vez do `df` original para todos os cálculos e gerações de gráficos. Devido ao modelo reativo do Streamlit, tudo se atualizará automaticamente a cada interação do usuário.

**Exemplo de Código Completo para `app.py` (Integração de Todos os Elementos)**:

````python
#```python
import streamlit as st
import pandas as pd
import plotly.express as px

# Carregar o dataset
try:
    df = pd.read_csv('usuarios_limpo.csv')
    df['data_cadastro'] = pd.to_datetime(df['data_cadastro'])
except FileNotFoundError:
    st.error("Arquivo 'usuarios_limpo.csv' não encontrado. Por favor, execute a célula de geração de dados sintéticos no Jupyter Notebook e salve o arquivo.")
    st.stop()

st.title("Dashboard de Análise de Usuários")
st.markdown("""
Este dashboard interativo permite explorar o perfil dos nossos usuários,
sua distribuição geográfica e a evolução dos cadastros ao longo do tempo.
Utilize os filtros na barra lateral para análises mais detalhadas.
""")

# Barra lateral para filtros
st.sidebar.header("Filtros")
estados_unicos = sorted(df['estado'].unique().tolist())
estados_selecionados = st.sidebar.multiselect(
    "Selecione o(s) Estado(s)",
    options=estados_unicos,
    default=estados_unicos
)

# Aplicar a lógica de filtragem
if estados_selecionados:
    df_filtrado = df[df['estado'].isin(estados_selecionados)]
else:
    df_filtrado = df
    st.warning("Nenhum estado selecionado. Exibindo dados de todos os estados.")

# Exibir KPIs
total_usuarios = df_filtrado['user_id'].nunique()
media_valor_compra = df_filtrado['valor_ultima_compra'].mean()

col1, col2 = st.columns(2)
with col1:
    st.metric(label="Total de Usuários", value=f"{total_usuarios:,}".replace(",", "."), delta="Simulado: +10%")
with col2:
    st.metric(label="Média Última Compra", value=f"R$ {media_valor_compra:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."), delta="Simulado: -2.5%")

st.markdown("---") # Separador visual

# Gráfico de barras: Distribuição de Usuários por Estado
st.subheader("Distribuição de Usuários por Estado")
usuarios_por_estado = df_filtrado['estado'].value_counts().reset_index()
usuarios_por_estado.columns = ['Estado', 'Número de Usuários']
fig_estado = px.bar(
    usuarios_por_estado,
    x='Estado',
    y='Número de Usuários',
    title='Distribuição de Usuários por Estado Selecionado',
    color='Número de Usuários',
    color_continuous_scale=px.colors.sequential.Plasma
)
st.plotly_chart(fig_estado, use_container_width=True)

st.markdown("---") # Separador visual

# Gráfico de linhas: Evolução dos Cadastros ao Longo do Tempo
st.subheader("Evolução dos Cadastros (por Mês)")
df_filtrado_copy_for_period = df_filtrado.copy() # Criar cópia para evitar SettingWithCopyWarning
df_filtrado_copy_for_period['mes_cadastro'] = df_filtrado_copy_for_period['data_cadastro'].dt.to_period('M').astype(str)
cadastros_por_mes = df_filtrado_copy_for_period.groupby('mes_cadastro').size().reset_index(name='Número de Cadastros')
cadastros_por_mes['mes_cadastro'] = pd.to_datetime(cadastros_por_mes['mes_cadastro']) # Converter de volta para datetime
cadastros_por_mes = cadastros_por_mes.sort_values('mes_cadastro')
fig_evolucao = px.line(
    cadastros_por_mes,
    x='mes_cadastro',
    y='Número de Cadastros',
    title='Evolução dos Cadastros ao Longo do Tempo Selecionado',
    markers=True
)
st.plotly_chart(fig_evolucao, use_container_width=True)

st.markdown("---") # Separador visual

# Exibir o DataFrame filtrado (opcional, para depuração/visualização de dados brutos)
st.subheader("Dados Filtrados (Amostra)")
st.dataframe(df_filtrado.head())
````

### 6. Organização do Layout do Dashboard

A organização visual de um dashboard é tão importante quanto a precisão dos dados que ele apresenta. Um layout bem estruturado facilita a compreensão e a navegação do usuário.

#### Utilizando `st.columns()` para Disposição Lado a Lado de KPIs e Gráficos

A função `st.columns()` do Streamlit permite dividir o layout da página em múltiplas colunas, o que é uma excelente maneira de organizar KPIs e gráficos lado a lado, melhorando significativamente a experiência visual do usuário. É possível especificar o número de colunas desejadas ou suas larguras relativas, proporcionando flexibilidade no design. Por exemplo, `st.columns(2)` cria duas colunas de largura igual, enquanto `st.columns([0.7, 0.3])` cria duas colunas com larguras proporcionais. A função `st.metric` é particularmente eficaz quando combinada com `st.columns`. Para manter uma boa aparência em todas as telas e seguir as melhores práticas de design, recomenda-se não aninhar colunas mais de uma vez.

#### Melhorando a Experiência Visual do Usuário

Um layout bem organizado não apenas melhora a estética, mas também reduz a carga cognitiva do usuário e guia seu olhar para as informações mais importantes. O uso de `st.columns` não serve apenas para acomodar mais elementos na tela, mas para criar um fluxo visual claro e direcionar a atenção do usuário. Isso se alinha com princípios de design de UX, como a colocação de informações cruciais no canto superior esquerdo da tela. Um layout desorganizado aumenta a carga cognitiva, enquanto um layout estruturado a minimiza, tornando o dashboard mais eficaz. Para desenvolvedores web, isso conecta diretamente suas habilidades de UI/UX com a construção de dashboards. `st.columns` oferece uma maneira Pythonica simples de aplicar esses princípios, resultando em dashboards mais profissionais e intuitivos.

### Próximos Passos: Publicação (Simulada) do Dashboard

Após a construção do dashboard, o próximo passo lógico é torná-lo acessível a outros. Embora a aula não inclua um deployment real, é importante discutir as opções para "publicar" (simular) o dashboard para que possa ser compartilhado.

As principais abordagens para o deployment de aplicações Streamlit incluem:

* **Streamlit Community Cloud**: Esta é a maneira mais direta de implantar aplicações Streamlit públicas gratuitamente. Requer que o código esteja em um repositório público do GitHub, juntamente com um arquivo `requirements.txt` listando as dependências. O processo é simplificado para "um clique" e a maioria das aplicações é implantada em poucos minutos.
* **Heroku**: Uma plataforma de nuvem que suporta a implantação de aplicações Python. Requer a criação de arquivos `requirements.txt`, `setup.sh` (para configuração do ambiente Streamlit) e `Procfile` (para instruir o Heroku sobre como executar a aplicação).
* **Contêiner Docker em um Servidor Web**: Para maior controle sobre o ambiente e dependências, a aplicação Streamlit pode ser containerizada usando Docker. Isso envolve a criação de um `Dockerfile` que define a imagem do ambiente, incluindo as bibliotecas Python e quaisquer dependências do sistema operacional. A imagem Docker pode então ser implantada em qualquer servidor que suporte Docker.

Encoraja-se os participantes a explorar essas opções de deployment e a adicionar mais funcionalidades aos seus dashboards, como a inclusão de mais métricas, diferentes tipos de gráficos, ou até mesmo a integração com fontes de dados em tempo real.
